```python
import numpy as np
```

```python
numeric_cols = df.select_dtypes(include=[np.number])
corr_matrix = numeric_cols.corr()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig = plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix Heatmap')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# Plot 1: Runs vs Balls Faced
sns.scatterplot(
    data=df, x='BF', y='Runs', hue='Ave', palette='viridis', ax=axes[0]
)
axes[0].set_title('Runs vs Balls Faced (BF)')

# Plot 2: Runs vs Centuries
sns.scatterplot(data=df, x='100', y='Runs', color='crimson', ax=axes[1])
axes[1].set_title('Runs vs Centuries (100s)')

# Plot 3: Batting Average vs Strike Rate
sns.scatterplot(data=df, x='SR', y='Ave', color='darkblue', ax=axes[2])
axes[2].set_title('Batting Average vs Strike Rate')

plt.tight_layout()
plt.show()
```

```python

sns.histplot(df['Ave'], kde=True, color='skyblue')
plt.title('Distribution of Batting Average')
plt.xlabel('Average')
plt.show()
```

```python
sns.histplot(df['SR'], kde=True, color='salmon')
plt.title('Distribution of Strike Rate')
plt.xlabel('Strike Rate')
plt.show()
```

```python
sns.boxplot(data=df[['Runs', 'BF']])
plt.title('Outliers in Runs & Balls Faced')
plt.show()
```

```python
sns.boxplot(data=df[['100', '50', '0']])
plt.title('Outliers in Centuries, Fifties & Ducks')
plt.show()
```
