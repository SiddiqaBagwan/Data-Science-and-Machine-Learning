```python
df.fillna("unknown",axis=1,inplace=True)
```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")
```

```python
df["added_year"] = df["date_added"].dt.year
df["added_month"] = df["date_added"].dt.month
df["added_day"] = df["date_added"].dt.day
```

```python
df["added_year"] = df["added_year"].fillna(0).astype(int)
df["added_month"] = df["added_month"].fillna(0).astype(int)
df["added_day"] = df["added_day"].fillna(0).astype(int)
```

```python
df.drop("date_added", axis=1, inplace=True)
```

```python
df.head(3)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encoder = LabelEncoder()

df["type_encoded"] = encoder.fit_transform(df["type"])
```

```python
df.head(3)
```

```python
df_onehot = pd.get_dummies(df, columns=["rating"])

print(df_onehot.head())
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()

cols = ["release_year", "added_year", "added_month", "added_day"]

df[cols] = scaler.fit_transform(df[cols])
```
