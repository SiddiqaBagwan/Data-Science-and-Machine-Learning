```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("odb1.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
top_players = df.nlargest(10, 'Runs')

plt.figure(figsize=(10,5))
sns.barplot(x='Player', y='Runs', data=top_players)

plt.xticks(rotation=45)
plt.title("Top 10 Players by Runs")
plt.xlabel("Player")
plt.ylabel("Runs")
plt.show()
```

```python
totals = df[['100', '50', '0']].sum()

print(totals)
```

```python
plt.figure(figsize=(7,7))

plt.pie(totals, labels=totals.index, autopct='%1.1f%%')

plt.title("Distribution of Centuries, Half-Centuries and Ducks")
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Ave'], bins=15)

plt.title("Distribution of Batting Average")
plt.xlabel("Batting Average")
plt.ylabel("Number of Players")
plt.show()
```

```python
numeric_data = df.select_dtypes(include='number')
```

```python
numeric_data = numeric_data.drop(columns=['Unnamed: 0'])
```

```python
correlation = numeric_data.corr()
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(correlation, annot=True, fmt='.2f', cmap='coolwarm')

plt.title("Correlation Heatmap")
plt.show()
```
